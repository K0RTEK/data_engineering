```python
def check_license_plate(license_plate: str) -> str:
    letters = set('АВЕКМНОРСТУХ')
    
    if len(license_plate) < 8 or len(license_plate) > 9:
        return "Номер не валиден."
    
    region_digits = 2 if len(license_plate) == 8 else 3
    
    first_letter = license_plate[0]
    three_numbers = license_plate[1:4]
    two_letters = license_plate[4:6]
    region = license_plate[6:6 + region_digits]
    
    if not (first_letter.isalpha() and three_numbers.isdigit() and 
            two_letters.isalpha() and region.isdigit()):
        return "Номер не валиден."
    
    if (first_letter not in letters or 
        any(letter not in letters for letter in two_letters)):
        return "Номер не валиден."
    
    number_part = license_plate[:6]
    return f"Номер {number_part} валиден. Регион: {region}."
```

```python
print(check_license_plate('А222ВС96'))
print(check_license_plate('AB22BB193')) 
```
