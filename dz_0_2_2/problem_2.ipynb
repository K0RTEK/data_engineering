```python
queries = [
    'смотреть сериалы онлайн',
    'новости спорта',
    'афиша кино',
    'курс доллара',
    'сериалы этим летом',
    'курс по питону',
    'серии про спорт'
]

total_queries = len(queries)
word_count = {}

for query in queries:
    words = query.split()
    word_num = len(words)
    word_count[word_num] = word_count.get(word_num, 0) + 1

for count in sorted(word_count.keys()):
    percentage = (word_count[count] / total_queries) * 100
    print(f"Поисковых запросов, содержащих {count} слов(а): {percentage:.2f}%")
```
