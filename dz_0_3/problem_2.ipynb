```python
# База данных документов (из Задания 1)
documents = [
    {"type": "passport", "number": "2207 876234", "name": "Василий Гупкин"},
    {"type": "invoice", "number": "11-2", "name": "Геннадий Покемонов"},
    {"type": "insurance", "number": "10006", "name": "Аристарх Павлов"}
]

directories = {
    '1': ['2207 876234', '11-2'],
    '2': ['10006'],
    '3': []
}

def find_owner_by_number(doc_number):
    """Находит владельца документа по его номеру."""
    for doc in documents:
        if doc["number"] == doc_number:
            return doc["name"]
    return None

def find_shelf_by_number(doc_number):
    """Находит полку, на которой хранится документ по его номеру."""
    for shelf, docs in directories.items():
        if doc_number in docs:
            return shelf
    return None

def main():
    while True:
        command = input("Введите команду: ").strip().lower()
        if command == 'p':
            doc_number = input("Введите номер документа: ").strip()
            owner = find_owner_by_number(doc_number)
            if owner:
                print(f"Результат:\nВладелец документа: {owner}")
            else:
                print("Результат:\nДокумент с таким номером не найден.")
        elif command == 's':
            doc_number = input("Введите номер документа: ").strip()
            shelf = find_shelf_by_number(doc_number)
            if shelf:
                print(f"Результат:\nДокумент хранится на полке: {shelf}")
            else:
                print("Результат:\nДокумент с таким номером не найден на полках.")
        elif command == 'q':
            print("Выход из программы.")
            break
        else:
            print("Неизвестная команда. Введите 'p' для поиска владельца, 's' для поиска полки или 'q' для выхода.")
```

```python
if __name__ == "__main__":
    main()
```
